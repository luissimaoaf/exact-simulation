## Live demo: a fresh exact skeleton (seed 3)

```python
import numpy as np
from exactjd.models.ou import OU
from exactjd.models.jump import QuadraticRateJump
from exactjd.jump.records import OUBridgeFactory
from exactjd.jump.skeleton import simulate_skeleton
ou = OU.isotropic(1.0, np.eye(2)); jump = QuadraticRateJump(2, 1.0, 1.0, 0.5)
sk = simulate_skeleton(ou, jump, OUBridgeFactory(ou), np.array([0.5, 0.5]), 1.0, np.random.default_rng(3))
print(sk.n_events, [round(e.tau, 4) for e in sk.events], sk.terminal)
print('refined radius', sk.refine(0.1).radius)
```
